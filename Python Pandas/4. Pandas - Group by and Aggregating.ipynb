# Pandas - Group by and Aggregating

```python
# importing pandas library

import pandas as pd
```

```python
# reading csv file

df = pd.read_csv("https://raw.githubusercontent.com/lncgthb/Python-Learnings/refs/heads/main/Python%20Pandas/Datasets/3.%20Flavors.csv")
df
```

```python
# grouping by "base flavor" and identifying the average

group_by_frame = df.groupby('Base Flavor')
```

```python
group_by_frame.mean(numeric_only = True)  # python update - we need to put numeric_only = True on the function MEAN
```

```python
# or aggreagate within the groupby (average or mean)

df.groupby("Base Flavor").mean(numeric_only = True)
```

```python
# or aggreagate within the groupby (count)

df.groupby("Base Flavor").count()
```

```python
# or aggreagate within the groupby (min)

df.groupby("Base Flavor").min()
```

```python
# or aggreagate within the groupby (max)

df.groupby("Base Flavor").max()
```

```python
# or aggreagate within the groupby (sum)

sum1 = df.groupby("Base Flavor").sum()


# the updated python summed the string so we need to drop string column datas

sum1.drop(columns = ["Flavor", "Liked"])
```

```python
# official "aggragate function" by using "dictionaries" with "list" in it

df.groupby("Base Flavor").agg({"Flavor Rating" : ["mean", "max", "count", "sum"], "Texture Rating" : ["mean", "max", "count", "sum"]})
```

```python
# GROUP ON MULTIPLE COLUMNS
df
```

```python
# GROUP ON MULTIPLE COLUMNS and identifying "mean"

df.groupby(["Base Flavor", "Liked"]).mean(numeric_only = True)
```

```python
# GROUP ON MULTIPLE COLUMNS and using aggregate function with dictionary with a list

df.groupby(["Base Flavor", "Liked"]).agg({"Flavor Rating" : ["mean", "max", "count", "sum"]})
```

```python
# shortcut generalize function

df.groupby("Base Flavor").describe()
```
