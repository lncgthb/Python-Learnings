# Pandas - Visualizing Data

```python
# importing libraries

import pandas as pd
import numpy as np
import matplotlib as plt
```

```python
# read/open dataframe

df = pd.read_csv("https://raw.githubusercontent.com/lncgthb/Python-Learnings/refs/heads/main/Python%20Pandas/Datasets/5.%20Ice%20Cream%20Ratings.csv")
df
```

```python
# change the index to "Date"

df = df.set_index("Date")
df
```

```python
# changing the style of the plot/chart before using it
import matplotlib.style

print(plt.style.available)
plt.style.use("seaborn-v0_8-talk")
```

```python
# visualizing through line chart

df.plot(kind = "line")
```

```python
# visualizing through line chart
# with sublplots = different columns

df.plot(kind = "line", subplots = True)
```

```python
# visualizing through line chart
# with different arguments -- inside the parenthesis ()

df.plot(kind = "line", title = "Ice Cream Ratings", xlabel = "Daily Ratings", ylabel = "Scores")
```

```python
# visualizing through bar chart
# with different arguments -- inside the parenthesis ()

df.plot(kind = "bar")
```

```python
# visualizing through bar chart
# with different arguments -- inside the parenthesis ()

df.plot(kind = "bar", stacked = True)
```

```python
# visualizing through bar chart
# with different arguments -- inside the parenthesis ()
# specific column only

df["Flavor Rating"].plot(kind = "bar")
```

```python
# visualizing through bar chart - horizontal
# with different arguments -- inside the parenthesis ()

df.plot.barh(stacked = True)
```

```python
# visualizing through scatter plot chart
# with different arguments -- inside the parenthesis ()

df.plot.scatter(x = "Texture Rating", y = "Overall Rating", s = 400, c = "Red")  # s=size --- c=color
```

```python
# visualizing through histogram plot chart
# with different arguments -- inside the parenthesis ()

df.plot.hist(bins = 10)
```

```python
# visualizing through box plot chart
# with different arguments -- inside the parenthesis ()

df.boxplot()
```

```python
# visualizing through area plot chart
# with different arguments -- inside the parenthesis ()

df.plot.area(figsize = (10, 5))
```

```python
# visualizing through pie plot chart
# with different arguments -- inside the parenthesis ()

df.plot.pie(y = "Flavor Rating", figsize = (10, 10))
```
