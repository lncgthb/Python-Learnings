# Pandas - Merging, Join, Concatenate - DataFrames

```python
# importing pandas library

import pandas as pd
```

```python
# reading/opening csv files as a dataframe

df1 = pd.read_csv("https://raw.githubusercontent.com/lncgthb/Python-Learnings/refs/heads/main/Python%20Pandas/Datasets/4.0.%20LOTR.csv")
df1
```

```python
# reading/opening csv files as a dataframe

df2 = pd.read_csv("https://raw.githubusercontent.com/lncgthb/Python-Learnings/refs/heads/main/Python%20Pandas/Datasets/4.1.%20LOTR%202.csv")
df2
```

```python
# MERGE

# default inner join - merge
df1.merge(df2)
```

```python
# MERGE

# actual way of merging inner join
df1.merge(df2, how = "inner", on = ["FellowshipID"])
```

```python
# MERGE

# actual way of merging inner join -- if you want to specify or joining 2 same data column
df1.merge(df2, how = "inner", on = ["FellowshipID", "FirstName"])
```

```python
# MERGE

# outer join - merge

df1.merge(df2, how = "outer")
```

```python
# MERGE

# left join - merge

df1.merge(df2, how = "left")
```

```python
# MERGE

# right join - merge

df1.merge(df2, how = "right")
```

```python
# MERGE

# cross join - merge   # but this is not use in actual work

df1.merge(df2, how = "cross")
```

```python
# JOIN - is better on working with INDEXES while the MERGE is for columns

# inner join - JOIN
# this will be diffuclt because it has so much work and code to do but we need to learn this
df1.join(df2, on = "FellowshipID", how = "outer", lsuffix = "_Left", rsuffix = "_Right")
```

```python
# JOIN - is better on working with INDEXES while the MERGE is for columns

# correct way to JOIN properly - but the FellowshipID is actually an index compare to the merge
# inner join
df4 = df1.set_index("FellowshipID").join(df2.set_index("FellowshipID"), lsuffix = "_Left", rsuffix = "_Right")
df4
```

```python
# JOIN - is better on working with INDEXES why the MERGE is for columns

# correct way to JOIN properly - but the FellowshipID is actually an index compare to the merge
# outer join
df4 = df1.set_index("FellowshipID").join(df2.set_index("FellowshipID"), lsuffix = "_Left", rsuffix = "_Right", how = "outer")
df4
```

```python
# CONCATENATE - instead of adding it to a similar data, concatenate is just adding df1 to df2 without changing anything

pd.concat([df1, df2])
```

```python
# CONCATENATE - instead of adding it to a similar data, concatenate is just adding df1 to df2 without changing anything

# doing inner JOIN
pd.concat([df1, df2], join = "inner")
```

```python
# CONCATENATE - instead of adding it to a similar data, concatenate is just adding df1 to df2 without changing anything

# doing outer JOIN
pd.concat([df1, df2], join = "outer")
```

```python
# CONCATENATE - instead of adding it to a similar data, concatenate is just adding df1 to df2 without changing anything

# doing join but in a axis
pd.concat([df1, df2], join = "outer", axis = 1)
```
